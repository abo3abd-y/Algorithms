## Insertion-sort Algoritm

We provide the insertion sort algorithm here.

```python
def insertion_sort(arr):

    if len(arr) < 2:
        return arr
    
    for j in range(1, len(arr)):
        key = arr[j]
        i = j - 1
        while i >= 0 and arr[i] > key:
            arr[i + 1] = arr[i]
            i -= 1
        arr[i + 1] = key
    return arr
```

```python
print(insertion_sort([5]))
```

```python
print(insertion_sort([1, 2, 3, 4, 5, 6]))
```

```python
print(insertion_sort([6, 5, 4, 3, 2, 1]))
```

```python
print(insertion_sort([3, 1, 4, 1, 5, 9, 2]))
```

```python
print(insertion_sort([7, 7, 7, 7, 7]))
```

```python
print(insertion_sort([2,1]))
print(insertion_sort([3,2,1]))
```

```python
print(insertion_sort([]))
```
